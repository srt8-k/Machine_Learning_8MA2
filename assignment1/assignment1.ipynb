```python
plt.figure(figsize=(8,5))
sns.histplot(df["Price (in USD)"], bins=20, kde=True)
plt.title("Distribution of Sports Car Prices")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Horsepower", y="Price (in USD)", data=df)
plt.title("Horsepower vs Price")
plt.show()
```
